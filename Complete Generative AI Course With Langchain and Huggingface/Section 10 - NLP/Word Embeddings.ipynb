# 📘 Word Embeddings – Cheatsheet

## 🔎 Core Idea
- **Word Embeddings** = dense vectors capturing **semantic meaning** + **contextual similarity**.
- Move from **sparse** (OHE, BoW, TF‑IDF) → **dense** (Word2Vec, GloVe, FastText).

---

## ✨ Key Properties
- **Dimensionality**: 100–300 (vs vocab size in OHE).
- **Semantic closeness**: Similar words → close vectors.
- **Arithmetic**:  
  \[
  \text{king} - \text{man} + \text{woman} \approx \text{queen}
  \]

---

## 📊 Representations

### Count / Frequency‑based
- **OHE** → Sparse, no meaning.  
- **BoW** → Counts only.  
- **TF‑IDF** → Weights by rarity, still context‑agnostic.

### Deep Learning‑based
- **Word2Vec**  
  - **CBOW**: Predict word from context.  
    - ✅ Fast, good for frequent words.  
    - ⚡ Best for **small datasets**.  
  - **Skip‑Gram**: Predict context from word.  
    - ✅ Good for rare words, fine semantics.  
    - ⚡ Best for **large datasets**.  
- **GloVe** → Matrix factorization + co‑occurrence stats.  
- **FastText** → Subword n‑grams → handles rare words, misspellings.

---

## 🧠 Word2Vec Architecture (Mini Diagrams)

### CBOW
```
Context Words → [Hidden Layer] → Target Word
```
Example:  
Input: ["the", "cat", "on"] → Predict: "mat"

### Skip‑Gram
```
Target Word → [Hidden Layer] → Context Words
```
Example:  
Input: "cat" → Predict: ["the", "on", "mat"]

---

## ⚙️ Training Tricks
- **Negative Sampling** → Efficient training.  
- **Hierarchical Softmax** → Faster probability computation.  
- **Improve CBOW/Skip‑Gram**:  
  - Increase **window size** → capture broader context.  
  - Use **CBOW** for small datasets, **Skip‑Gram** for large.

---

## 🚀 Applications
- Text classification (spam, sentiment).  
- Machine translation.  
- Semantic search.  
- Chatbots.  
- Recommendations.

---

## ⚖️ Pros & Cons

### ✅ Advantages
- Sparse → Dense.  
- Captures semantic info.  
- Fixed dimensions regardless of vocab size.  
- Handles OOV (FastText).  

### ❌ Limitations
- Static → no polysemy handling.  
- Context‑independent.  
- Modern contextual embeddings (ELMo, BERT, GPT) solve this.

---

## 📐 Variants
- **Word2Vec** → vector per word.  
- **Avg Word2Vec** → average of word vectors for sentence/document.

---

## 📚 Libraries & Resources
- **Gensim** → Word2Vec, FastText.  
- **Pre‑trained Google Word2Vec** (300‑dim, 3M words).  
- **GloVe** → Stanford pre‑trained vectors.  

---

## 📝 Quick Decision Guide
| Dataset Size | Best Method | Notes |
|--------------|-------------|-------|
| Small        | CBOW        | Fast, frequent words |
| Large        | Skip‑Gram   | Rare words, fine semantics |
| Rare words / OOV | FastText | Subword n‑grams |
| Global stats | GloVe       | Co‑occurrence matrix |

---

## 🔗 Similarity Measure
- **Cosine Similarity**:  
  - Range: -1 (opposite) → 1 (identical).  
  - 0 = orthogonal (no similarity).  

---

# 🎯 Final Cheat Pointers
- Use **CBOW** for speed, **Skip‑Gram** for accuracy.  
- Increase **window size** for richer context.  
- For rare words → **FastText**.  
- For global corpus stats → **GloVe**.  
- For contextual meaning → **BERT/ELMo/GPT**.
