```python
import nltk 
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')
```

```python
lemmatizer = WordNetLemmatizer()

word_list = [
    "running",
    "ran",
    "runs",
    "am",
    "are",
    "is",
    "was",
    "were",
    "better",
    "best",
    "feet",
    "mice",
    "studies",
    "studying",
    "applied",
    "meeting",
    "cats",
    "children",
    "thought",
    "gone",
    "books",
    "histories",
    "women",
    "leaves"
]
```

```python
for word in word_list:
    print(word + " -----> " + lemmatizer.lemmatize(word, pos='v'))
```
