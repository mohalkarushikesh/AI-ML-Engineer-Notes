````python
import nltk
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
````

````python
data = {
    'emp id': [10, 20, 30, 40, 50],
    'gender': ['M', 'F', 'F', 'M', 'F'],
    'remarks': ['Good', 'Nice', 'Good', 'Great', 'Nice']
}
````

````python
df = pd.DataFrame(data)
print(f"Original Employee Data: \n{df}\n")
````

````python
# One-Hot Encoding 
pandas_df_encoded = pd.get_dummies(df, columns=['gender', 'remarks'], drop_first=True)
print(f"One Hot Encoding data using pandas: \n{pandas_df_encoded}\n")
````

````python
# Using Scikit Learn for One-Hot Encoding

# Print the original dataframe
print(f"Original dataframe: \n{df}\n")

# Select all categorical columns (dtype = object) from the dataframe
categorical_columns = df.select_dtypes(include=['object']).columns.tolist()

# Initialize the OneHotEncoder
# sparse_output=False ensures the output is a NumPy array instead of a sparse matrix
encoder = OneHotEncoder(sparse_output=False)

# Fit the encoder on categorical columns and transform them into one-hot encoded values
one_hot_encoded = encoder.fit_transform(df[categorical_columns])

# Convert the one-hot encoded NumPy array into a DataFrame
# Use encoder.get_feature_names_out() to generate proper column names
one_hot_df = pd.DataFrame(one_hot_encoded, columns=encoder.get_feature_names_out(categorical_columns))

# Concatenate the original dataframe with the new one-hot encoded dataframe
df_encoded = pd.concat([df, one_hot_df], axis=1)

# Drop the original categorical columns since they are now encoded
df_encoded = df_encoded.drop(categorical_columns, axis=1)

# Print the final dataframe after one-hot encoding
print(f"One hot encoding using sklearn: \n{df_encoded}\n")
````

- Convert the one-hot encoded NumPy array into a DataFrame
- Use encoder.get_feature_names_out() to generate proper column names
```
one_hot_df = pd.DataFrame(one_hot_encoded, columns=encoder.get_feature_names_out(categorical_columns))
```
👉 Why concatenate?
- We join the original dataframe with the new one-hot encoded dataframe
- so that numeric columns (like 'emp id') remain alongside the encoded categorical columns.
- This way, the final dataframe contains both the original numeric data
- and the transformed categorical data in one place.

- Concatenate the original dataframe with the new one-hot encoded dataframe
```

```
