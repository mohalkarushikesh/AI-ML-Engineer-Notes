```python
import nltk 
from nltk import trigrams
from nltk.corpus import reuters
from collections import defaultdict
```

```python
nltk.download('reuters')
nltk.download('punkt')
```

```python

# Tokenize the Reuters corpus into words
words = list(reuters.words())
tri_grams = list(trigrams(words))

# Build trigram model: counts of next word given previous two
model = defaultdict(lambda: defaultdict(lambda: 0))

for w1, w2, w3 in tri_grams: 
    model[(w1, w2)][w3] += 1 

# Normalize counts into probabilities
for w1_w2 in model:  # iterate over bigram keys
    total_count = float(sum(model[w1_w2].values()))
    for w3 in model[w1_w2]:
        model[w1_w2][w3] /= total_count

# Prediction function
def predict_next_word(w1, w2):
    next_word_probs = model[(w1, w2)]
    if next_word_probs:
        return max(next_word_probs, key=next_word_probs.get)
    else:
        return "No prediction available"

# Example usage
print("The stock", predict_next_word('The', 'stock'))
```

```python
# Example usage
print("The possible", predict_next_word('The', 'possible'))
```
