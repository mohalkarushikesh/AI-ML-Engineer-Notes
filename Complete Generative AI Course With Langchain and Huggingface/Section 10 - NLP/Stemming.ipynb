```python
print(nltk.__version__)
```

```python
from nltk import SnowballStemmer
```

```python
sb_stem = SnowballStemmer("english")
print(sb_stem.stem("running"))
```

```python
# porterstemmer 

from nltk import PorterStemmer

pr_stemer = PorterStemmer()
print(pr_stemer.stem("singing"))
```

```python
# RegexpStemmer

from nltk import RegexpStemmer

regexp_s = RegexpStemmer('ing$')

print(regexp_s.stem("Reading"))
```
