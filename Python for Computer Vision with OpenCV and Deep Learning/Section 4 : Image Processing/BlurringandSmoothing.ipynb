```python
# Step 1: Load the puppy image
puppy_img = cv2.imread('../puppy.jpg', 0)  # Load in grayscale

# Check if image loaded successfully
if puppy_img is None:
    print("Error: Could not load puppy.jpg. Trying alternative path...")
    puppy_img = cv2.imread('../puppy.jpg', 0)  # Try current directory

if puppy_img is not None:
    print(f"Puppy image loaded successfully! Shape: {puppy_img.shape}")
    print(f"Image data type: {puppy_img.dtype}")
    print(f"Min pixel value: {puppy_img.min()}")
    print(f"Max pixel value: {puppy_img.max()}")
else:
    print("Error: Could not load puppy image. Please check the file path.")
```

```python
# Step 2: Add noise to the image
if puppy_img is not None:
    # Add salt and pepper noise
    noisy_img = puppy_img.copy()
    
    # Add salt noise (white pixels)
    salt_noise = np.random.random(puppy_img.shape) < 0.05  # 5% of pixels
    noisy_img[salt_noise] = 255
    
    # Add pepper noise (black pixels)
    pepper_noise = np.random.random(puppy_img.shape) < 0.05  # 5% of pixels
    noisy_img[pepper_noise] = 0
    
    print("Salt and pepper noise added successfully!")
    print(f"Noisy image shape: {noisy_img.shape}")
else:
    print("Cannot add noise - puppy image not loaded")
```

```python
# Load image and add noise
img = cv2.imread('../puppy.jpg', 0)
if img is None:
    img = cv2.imread('puppy.jpg', 0)

if img is not None:
    # Add salt and pepper noise
    noisy = img.copy()
    salt_noise = np.random.random(img.shape) < 0.1  # 10% noise
    pepper_noise = np.random.random(img.shape) < 0.1
    noisy[salt_noise] = 255
    noisy[pepper_noise] = 0
    
    # Show noisy image
    plt.figure(figsize=(10, 5))
    plt.subplot(1, 2, 1)
    plt.imshow(noisy, cmap='gray')
    plt.title('Noisy Image', fontsize=14)
    plt.xticks([])
    plt.yticks([])
    
    # Apply median blur to reduce noise
    denoised = cv2.medianBlur(noisy, 5)
    
    # Show denoised image
    plt.subplot(1, 2, 2)
    plt.imshow(denoised, cmap='gray')
    plt.title('After Median Blur', fontsize=14)
    plt.xticks([])
    plt.yticks([])
    
    plt.tight_layout()
    plt.show()
    
    print("✅ Noise reduction complete!")
    print(f"Original std: {img.std():.1f}")
    print(f"Noisy std: {noisy.std():.1f}")
    print(f"Denoised std: {denoised.std():.1f}")
else:
    print("❌ Could not load image")
```
